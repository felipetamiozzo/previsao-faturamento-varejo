Maio possui o maior volume de vendas, destacando-se dos demais meses. Há variações significativas no volume ao longo do ano.

```python
df.groupby('mes_num')['VUF_QTPRODUTO'].sum().plot(kind='bar')
plt.title('Quantidade de produtos vendidos por mês')
plt.show()
```

Setembro apresenta o maior volume de descontos. Os descontos variam ao longo dos meses, indicando mudanças na estratégia comercial.

```python
df.groupby('mes_num')['VUF_VLRDESCONTO'].sum().plot(kind='bar')
plt.title('Descontos por mês')
plt.show()
```
